Abrindo o csv notei que o nome "Robert Downey, Jr." estava separado por vírgula. Mas o separador é vírgula também, o que causava erro na hora de usar o dataset. Se eu mudasse o separador para ' ; ' não iria resolver por que "Robert Downey, Jr." passaria para "Robert Downey; Jr.", o que geraria o mesmo erro.

Então optei por abrir com separador ' , ' e posteriormente concatenar "Robert Downey" e "Jr." ficando "Robert Downey Jr."


```python
with open('/content/actors.csv', 'r') as arquivo:
    linhas = arquivo.readlines()

dataset = []
for linha in linhas:
    # Dividindo cada linha por vírgula para obter os campos individuais
    campos = linha.strip().split(',')
    # Encontrando o nome "Robert Downey" e o sobrenome é "Jr."
    if campos[0] == "Robert Downey" and campos[1].strip() == "Jr.":
        # Concatenando o nome e o sobrenome
        campos[0] = campos[0] + " " + campos[1].strip()
        # Removendo o sobrenome da lista de campos
        campos.pop(1)
    dataset.append(campos)

for linha in dataset:
    print(linha)
```

## Etapa 2

Apresente a média de receita de bilheteria bruta dos principais filmes, considerando todos os atores. Estamos falando da média da coluna **'Gross'**.

```python
def avg_gross(dataset):
  # Iniciando as variáveis
  total_gross = 0
  total_actors = 0
  # Lendo o dataset linha por linha e eliminando a primeira
  for linha in dataset[1:]:
    # Variáveis temporárias
    gross = float(linha[5])
    total_gross += gross
    total_actors += 1
  # encontrando a média de receita bruta dos filmes considerando todos os atores
  total_avg_gross = total_gross / total_actors
  return (total_avg_gross)

# Imprimindo os resultados
print(f'A média da receita de bilheteria bruta é {avg_gross(dataset):.2f}')
```

## Etapa 4

A coluna **#1 Movie** contém o filme de maior bilheteria em que o ator atuou. Realize a contagem de aparições destes filmes no dataset, listando-os ordenados pela quantidade de vezes e que estão presentes. Considere a ordem decrescente e, em segundo nível, o nome do filme.

Ao escrever no arquivo, considere o padrão de saída **(sequência) - O filme (nome do filme) aparece (quantidade) vez(es) no dataset**, adicionando um resultado a cada linha.

```python
def count_movies(dataset):
    movie_counts = {}
    # Lendo o dataset linha por linha e eliminando a primeira
    for linha in dataset[1:]:
        # A variável temporária
        filme = linha[4]
        # Se o filme já estiver no dicionário, incrementa a contagem
        if filme in movie_counts:
            movie_counts[filme] += 1
        # Se o filme não estiver no dicionário, adiciona com a contagem 1
        else:
            movie_counts[filme] = 1

    # Ordenando o dicionário pela contagem dos filmes em ordem decrescente e, em caso de empate, pelo nome do filme
    sorted_movies = sorted(movie_counts.items(), key=lambda x: (-x[1], x[0]))

    return sorted_movies

# Imprimindo os resultados
sorted_movies = count_movies(dataset)
for i, (filme, contagem) in enumerate(sorted_movies, 1):
    if contagem > 1:
        print(f'{i}. O filme {filme} aparece {contagem} vezes no dataset.')
    else:
        print(f'{i} - O filme {filme} aparece {contagem} vez no dataset.')
```

## Etapa 5

Apresente a lista dos atores ordenada pela receita bruta de bilheteria de seus filmes (coluna **'Total Gross'**), em ordem decrescente.

Ao escrever no arquivo considere o padrão de saída **(nome do ator) - (receita total bruta)**, adicionando um resultado a cada linha.

```python
def list_actors_gross(dataset):
    # Criando uma lista para armazenar os atores e suas receitas brutas totais
    actors_gross = []
    # Lendo o dataset linha por linha e eliminando a primeira
    for linha in dataset[1:]:
        # Variáveis temporárias
        ator = linha[0]
        total_gross = float(linha[1])
        # Adicionando o ator e sua receita bruta total à lista
        actors_gross.append((ator, total_gross))

    # Ordenando a lista pela receita bruta total em ordem decrescente
    sorted_actors = sorted(actors_gross, key=lambda x: -x[1])

    return sorted_actors

# Imprimindo os resultados
sorted_actors = list_actors_gross(dataset)
for ator, total_gross in sorted_actors:
    print(f'{ator} - {total_gross}')
```
