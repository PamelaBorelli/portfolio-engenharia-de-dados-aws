## Levantando informações sobre o csv movies.csv

```python
movies = pd.read_csv('movies.csv', sep='|')
movies.head(2)
```

```python
movies.info()
```

```python
# Observando o ano de produção dos filmes
ano_lancamento_movies = movies['anoLancamento'].unique()
ano_lancamento_movies
```

```python
# Quais gêneros estão contidos no dataset
genero_movies = movies['genero'].unique()
genero_movies
```

```python
# Observando as médias de notas
bins = [i/10 for i in range(0, 101, 10)]
labels = [f'{i}-{i+0.9}' for i in range(0, 10)]

movies['notas_agrupadas'] = pd.cut(movies['notaMedia'], bins=bins, labels=labels, include_lowest=True)

contagem = movies['notas_agrupadas'].value_counts().sort_index()
contagem
```

```python
# Filtrando o DataFrame para incluir apenas linhas onde o gênero contém 'war' ou 'crime'
movies_filtrados = movies[movies['genero'].str.contains('war|crime', case=False, na=False)]

quantidade = movies_filtrados['genero'].count()
print(f'Quantidade de Filmes que estão incluídos no gênero guerra ou crime: {quantidade}')

# Imprime parte do dataset filtrado
movies_filtrados
```

```python
# Conte o número de nomes de artistas únicos
num_artistas_unicos = movies['nomeArtista'].nunique()

# Filtrar as linhas onde o gênero do artista é 'ator'
movies_atores = movies[movies['generoArtista'] == 'actor']

# Filtrar as linhas onde o gênero do artista é 'atriz'
movies_atrizes = movies[movies['generoArtista'] == 'actress']

# Conte o número de atores únicos
num_atores_unicos = movies_atores['nomeArtista'].nunique()

# Conte o número de atrizes únicas
num_atrizes_unicas = movies_atrizes['nomeArtista'].nunique()

print(f'O número total de artistas únicos é: {num_artistas_unicos}')
print(f'Desse total, o número de atores únicos é: {num_atores_unicos}')
print(f'Desse total, o número de atrizes únicas é: {num_atrizes_unicas}')
```
