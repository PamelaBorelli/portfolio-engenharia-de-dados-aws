### Etapa 1

```python
# Importando as bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import findspark
findspark.init()
from pyspark.sql import SparkSession
from pyspark import SparkContext, SQLContext
from pyspark.sql.functions import col
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import rand, when, udf
import random
```

```python
# iniciando sessão Spark
spark = SparkSession.builder.appName("mySparkSession").getOrCreate()
```

### Etapa 2

```python
# Baixando o arquivo nomes_aleatórios.txt e lendo as primeiras linhas
df = spark.read.csv("/content/nomes_aleatórios.txt", header=False, inferSchema=True)

# Renomeia a coluna 1 para "nomes"
df = df.withColumnRenamed("_c0", "nomes")
df.show(5)
```

### Etapa 3

```python
# Incluindo coluna escolaridade
df = df.withColumn("Escolaridade", when(rand() < 0.33, "Fundamental")
                                    .when(rand() < 0.66, "Médio")
                                    .otherwise("Superior"))

# Exiba o DataFrame resultante
df.show(10)
```

### Etapa 4

```python
# Países da América do Sul
paises_sul_america = ["Brasil", "Argentina", "Chile", "Colômbia", "Equador", "Peru",
                      "Bolívia", "Paraguai", "Uruguai", "Venezuela", "Suriname", "Guiana",
                      "Guiana Francesa"]

# Criando uma função UDF para selecionar um país aleatoriamente
def random_country():
    return random.choice(paises_sul_america)

random_country_udf = udf(random_country, StringType())

# Adicionando a coluna "Pais" com valores aleatórios
df = df.withColumn("Pais", random_country_udf())
df.show()
```

### Etapa 5

```python
# Defindo intervalo de anos
ano_minimo = 1945
ano_maximo = 2010

# Criando uma função UDF para gerar um ano aleatório
def random_birth_year():
    return random.randint(ano_minimo, ano_maximo)

random_birth_year_udf = udf(random_birth_year, IntegerType())

# Adicionando a coluna "AnoNascimento" com valores aleatórios
df = df.withColumn("AnoNascimento", random_birth_year_udf())
df.show()
```

### Etapa 6

```python
# Filtrar as pessoas nascidas no século XXI
df_select = df.filter((df.AnoNascimento >= 2000) & (df.AnoNascimento <= 2010))

# Exibir os 10 primeiros nomes do DataFrame df_select
df_select.select("nomes").show(10)
```

### Etapa 7


```python
# Registre o DataFrame original como uma tabela temporária
df.createOrReplaceTempView("pessoas")

# Consulta SQL para selecionar as pessoas nascidas neste século
consulta_sql = """
    SELECT nomes
    FROM pessoas
    WHERE AnoNascimento BETWEEN 2000 AND 2010
"""

# Executa a consulta e armazena os resultados em df_select
df_select = spark.sql(consulta_sql)
df_select.show(10)
```

### Etapa 8

```python
# Filtrar pessoas nascidas entre 1980 e 1994
df_filtered = df.filter((df.AnoNascimento >= 1980) & (df.AnoNascimento <= 1994))

# Conta o número de pessoas
total_pessoas = df_filtered.count()
print(f"Número de pessoas nascidas entre 1980 e 1994 - Geração Millennials: {total_pessoas}")
```

### Etapa 9

```python
# Tabela temporária
df.createOrReplaceTempView("pessoas")

# Consulta SQL para contar o número de pessoas
consulta_sql = """
    SELECT COUNT(*) AS Millenials
    FROM pessoas
    WHERE AnoNascimento BETWEEN 1980 AND 1994
"""

# Executar a consulta
df_filtered = spark.sql(consulta_sql)
df_filtered.show()
```

### Etapa 10

```python
# Tabela temporária
df.createOrReplaceTempView("pessoas")

# Consulta SQL para contar o número de pessoas em cada geração
consulta_sql = """
    SELECT Pais,
           CASE
               WHEN AnoNascimento BETWEEN 1944 AND 1964 THEN 'Baby Boomers'
               WHEN AnoNascimento BETWEEN 1965 AND 1979 THEN 'Geração X'
               WHEN AnoNascimento BETWEEN 1980 AND 1994 THEN 'Millennials'
               WHEN AnoNascimento BETWEEN 1995 AND 2015 THEN 'Geração Z'
               ELSE 'Outra Geração'
           END AS Geracao,
           COUNT(*) AS Quantidade
    FROM pessoas
    GROUP BY Pais, Geracao
    ORDER BY Pais, Geracao
"""

# Executa a consulta e armazena os resultados em um novo DataFrame
df_resultado = spark.sql(consulta_sql)
df_resultado.show()
```
